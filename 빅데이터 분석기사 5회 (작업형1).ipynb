## 데이터 전처리 & 피처엔지니어링

```python
#결측치 없음, 이상치 없음, 숫자로다 바꿔주면 될듯?
target = train.pop('Attrition_Flag')
target_id = test.pop('CLIENTNUM')
df = pd.concat([train, test], axis = 0)
data_oh = pd.get_dummies(df)
train_oh = data_oh[:len(train)].copy()
test_oh = data_oh[len(train):].copy()
print(train.shape, test.shape, train_oh.shape, test_oh.shape)
```

## 검증 데이터 분리

```python
train_oh.shape, target.shape, test_oh.shape
```

```python
from sklearn.model_selection import train_test_split
X_train, X_val, y_train, y_val = train_test_split(train_oh, target, test_size = 0.2, random_state = 0)
X_train.shape, X_val.shape, y_train.shape, y_val.shape
```

## 모델 & 평가

```python
from sklearn.ensemble import RandomForestClassifier
rfcf = RandomForestClassifier()
rfcf.fit(X_train, y_train)
pred1 = rfcf.predict(X_val)
pred2 = rfcf.predict_proba(X_val)
pred1[:10]
pred2[:10]
```

```python
from sklearn.metrics import roc_auc_score, precision_score, recall_score, f1_score, accuracy_score
# 정확도
print(accuracy_score(y_val, pred1))

# 정밀도
print(precision_score(y_val, pred1))

# 재현율 (민감도)
print(recall_score(y_val, pred1))

# F1
print(f1_score(y_val, pred1))

# roc-auc
print(roc_auc_score(y_val, pred2[:,1]))
#베이스 라인
#0.9568167797655768
#0.927536231884058
#0.7773279352226721
#0.8458149779735683
#0.9868155861605643
```

```python
import lightgbm as lgb
lgbmc = lgb.LGBMClassifier(random_state = 0, verbose = -1)
lgbmc.fit(X_train, y_train)
pred1 = lgbmc.predict(X_val)
pred2 = lgbmc.predict_proba(X_val)
# 정확도
print(accuracy_score(y_val, pred1))

# 정밀도
print(precision_score(y_val, pred1))

# 재현율 (민감도)
print(recall_score(y_val, pred1))

# F1
print(f1_score(y_val, pred1))

# roc-auc
print(roc_auc_score(y_val, pred2[:,1]))
```
