## 작업형2 - 통신사에서 고객에게 청구될 총 금액을 예측하시오.
- 제공된 데이터 목록:
    - churn_train.csv(훈련 데이터)
    - churn_test.csv(평가용 데이터)
- 예측할 컬럼: TotalCharges(총 청구액)
- 학습용 데이터(churn_train.csv)를 이용하여 총 청구액을 예측하는 모델을 만든 후 이를 평가용 데이터(churn_test.csv)
에 적용해 얻은 예측값을 다음과 같은 형식의 CSV 파일로 생성하시오.

- 제출 파일은 다음 1개의 컬럼을 포함해야 한다.
    - pred: 예측된 총 청구액
    - 제출 파일명: ‘result.csv’
- 제출한 모델의 성능은 MAE(Mean Absolute Error) 평가지표에 따라 채점한다.


## 문제정의

```python
# 문제정의
# 평가: MAE
# target: TotalCharges
# 최종파일: result.csv(컬럼 1개 pred)
```

## 데이터 불러오기

```python
# 라이브러리 및 데이터 불러오기
import pandas as pd
train = pd.read_csv()
test = pd.read_csv()
```

## 탐색적 데이터 분석(EDA)

```python
# 데이터 샘플
train.head()
```

```python
# 데이터 크기
print("Train Shape:", train.shape)
print("Test Shape:", test.shape)
```

```python
# 데이터 정보(자료형)
print(train.info())
```

```python
# train 결측치 수
print(train.isnull().sum().sum())
```

```python
# test 결측치 수
print(test.isnull().sum().sum())
```

```python
# customerID unique 수
print(train['customerID'].nunique())
```

```python
# target 기술 통계
print(train['TotalCharges'].describe())
```

## 데이터 전처리

```python
# 데이터 전처리
train = train.drop('customerID', axis=1)
test = test.drop(['customerID'], axis=1)
target = train.pop('TotalCharges')
```

```python
# 레이블 인코딩
from sklearn.preprocessing import LabelEncoder
cols = train.select_dtypes(include='O').columns

for col in cols:
    le = LabelEncoder()
    train[col] = le.fit_transform(train[col])
    test[col] = le.transform(test[col])
```

```python
train.head()
```

## 검증데이터 나누기

```python
# 검증데이터 분리
from sklearn.model_selection import train_test_split
X_tr, X_val, y_tr, y_val = train_test_split(train, target, test_size=0.2, random_state=0)
X_tr.shape, X_val.shape, y_tr.shape, y_val.shape
```

## 모델 학습 및 평가

```python
# 랜덤포레스트
from sklearn.ensemble import RandomForestRegressor
rf = RandomForestRegressor(random_state=0)
rf.fit(X_tr, y_tr)
pred = rf.predict(X_val)
pred[:10]
```

```python
# MAE
from sklearn.metrics import mean_absolute_error
print(mean_absolute_error(y_val, pred))
# 951.0960435538027
```

```python
# LightGBM
import lightgbm as lgb
lg = lgb.LGBMRegressor(random_state=0, verbose=-1)
lg.fit(X_tr, y_tr)
pred = lg.predict(X_val)
print(mean_absolute_error(y_val, pred))
# 952.7925407798712
```

```python
# 최종 제출 파일
pred = rf.predict(test)
result = pd.DataFrame({
    'pred':pred
})
result.to_csv("result.csv", index=False)
```
